Bài 6

```python
import math

a, b, c = map(float, input("Nhập 3 cạnh a, b, c: ").split())
def dientich_heron(a, b, c):
  if a + b <= c or a + c <= b or b + c <= a:
    print("Ba cạnh này không tạo thành một tam giác!")
  p = (a + b + c) / 2
  s = math.sqrt(p * (p - a) * (p - b) * (p - c))
  return s
print(f"Diện tích tam giác: {dientich_heron(a, b, c):.1f}")
```
